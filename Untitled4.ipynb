```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from transformers import pipeline


# 1. TF-IDF Classifier Class
class TFIDFComplaintClassifier:

    def __init__(self):
        self.vectorizer = TfidfVectorizer(max_features=1000)
        self.model = LogisticRegression()
        self.is_trained = False

    def fit(self, X_train, y_train):
        X_vec = self.vectorizer.fit_transform(X_train)
        self.model.fit(X_vec, y_train)
        self.is_trained = True

    def predict(self, texts):
        if not self.is_trained:
            # Simple dummy fallback if not trained on dataset yet
            return ["unauthorized transaction" for _ in texts]
        X_vec = self.vectorizer.transform(texts)
        return self.model.predict(X_vec)


# 2. Zero-Shot Transformer Classifier Class
class TransformerComplaintClassifier:

    def __init__(self, model_name: str = "facebook/bart-large-mnli"):
        self.model_name = model_name
        self.clf = None

    def load(self):
        print(f"Loading transformer model: {self.model_name}...")
        self.clf = pipeline(
            "zero-shot-classification", model=self.model_name
        )
        print("Model loaded successfully!")

    def predict(self, texts, candidate_labels=None):
        if self.clf is None:
            self.load()

        if candidate_labels is None:
            candidate_labels = [
                "refund request",
                "unauthorized transaction",
                "card issue",
                "fee dispute",
            ]

        return self.clf(texts, candidate_labels)

    def compare_with_tfidf(self, tfidf_clf, test_texts, candidate_labels=None):
        if candidate_labels is None:
            candidate_labels = [
                "refund request",
                "unauthorized transaction",
                "card issue",
                "fee dispute",
            ]

        results = []
        for text in test_texts:
            tfidf_pred = (
                tfidf_clf.predict([text])[0]
                if hasattr(tfidf_clf, "predict")
                else "N/A"
            )
            trans_res = self.predict(text, candidate_labels)
            top_trans_label = trans_res["labels"][0]
            top_trans_score = trans_res["scores"][0]

            results.append({
                "Complaint Text": text,
                "TF-IDF Prediction": tfidf_pred,
                "Transformer Prediction": top_trans_label,
                "Confidence": f"{top_trans_score:.2%}",
            })

        return pd.DataFrame(results)


print("Classifiers ready to use!")
```

```python
# Initialize models
tfidf_clf = TFIDFComplaintClassifier()
transformer_clf = TransformerComplaintClassifier()

# Load BART transformer
transformer_clf.load()

# Test cases for Day 9 evaluation
eval_texts = [
    # Semantic Edge Cases where TF-IDF fails without exact keywords
    "Card was charged but item never arrived",
    "There is a mystery $45 deduction on my statement from yesterday",
    "My plastic card split down the middle and won't swipe",
    "Why was I hit with a monthly service cost on my free account?",
    "I sent money to the wrong account number by accident",
    "Merchant agreed to cancel the order but the charge is still pending",
    "Someone cloned my card details and bought gift cards",
    "The chip isn't reading at any chip reader terminals",
    "ATM swallowed my debit card and didn't give it back",
    "Please waive the late payment penalty fee for this billing cycle",
]

candidate_labels = [
    "refund request",
    "unauthorized transaction",
    "card issue",
    "fee dispute",
]

# Run zero-shot comparison table
df_results = transformer_clf.compare_with_tfidf(
    tfidf_clf, eval_texts, candidate_labels
)

# Display table
pd.set_option("display.max_colwidth", None)
df_results
```
